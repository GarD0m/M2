```python
import subprocess
import sys
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory(prefix="m2-") as temp:
    repo = Path(temp) / "M2"
    subprocess.run(["git", "clone", "--quiet", "--depth", "1", "https://github.com/GarD0m/M2.git", str(repo)], check=True)
    subprocess.run([sys.executable, "-m", "pip", "install", "--quiet", "-r", "requirements.txt"], cwd=repo, check=True)
    subprocess.run([sys.executable, "-m", "llm5.kaggle"], cwd=repo, check=True)
```
